```python
# Словарь бюджетов
from countries_data import countries

budget_levels = {
    "экономный": 1,
    "средний": 2,
    "высокий": 3
}


# Баллы за параметры
def calculate_score(info, user_month, user_climate, user_interests, user_rest):

    score = 0

    # Сезонность — 20 баллов
    if user_month in info["best_months"]:
        score += 20

    # Климат — 15 баллов
    if user_climate in info["climate"]:
        score += 15

    # Интересы — до 50 баллов
    matching_interests = set(user_interests) & set(info["interests"])

    if user_interests:
        score += 50 * len(matching_interests) / len(user_interests)

    # Тип отдыха — до 15 баллов
    country_rest = info["rest_type"]

    if user_rest == country_rest:
        score += 15
    elif user_rest == "смешанный" or country_rest == "смешанный":
        score += 10

    return round(score, 2)



# Проверка месяцев для поездки
def get_reasons(info, user_month, user_climate, user_interests, user_rest):

    reasons = []

    if user_month in info["best_months"]:
        reasons.append("подходящий сезон")

    if user_climate in info["climate"]:
        reasons.append("подходящий климат")

    matching_interests = [
        interest for interest in user_interests
        if interest in info["interests"]
    ]

    if matching_interests:
        reasons.append("интересы: " + ", ".join(matching_interests))

    if user_rest == info["rest_type"]:
        reasons.append("подходящий тип отдыха")
    elif user_rest == "смешанный" or info["rest_type"] == "смешанный":
        reasons.append("частично подходящий тип отдыха")

    return reasons


# Рекомендованные страны функция
def recommend_countries(user_budget, user_continent, user_month,
                        user_climate, user_interests, user_rest):

    results = []

    for name, info in countries.items():

        # Фильтр бюджета
        if budget_levels[info["budget"]] > budget_levels[user_budget]:
            continue

        # Фильтр континента
        country_continent = info["continent"].lower()

        if user_continent == "америка":
            if country_continent not in ["северная америка", "южная америка"]:
                continue

        elif user_continent != "любой" and country_continent != user_continent:
            continue

        # Рассчитываем баллы
        score = calculate_score(
            info, user_month, user_climate, user_interests, user_rest
        )

        # Получаем причины рекомендации
        reasons = get_reasons(
            info, user_month, user_climate, user_interests, user_rest
        )

        # Сохраняем результат
        results.append({
            "country": name,
            "score": score,
            "reasons": reasons
        })

    # Сортируем страны по баллам
    results = sorted(
        results,
        key=lambda item: item["score"],
        reverse=True
    )

    return results[:5]


# Получаем ответы пользователя
user_budget = input(
    "Бюджет (экономный, средний, высокий): "
).strip().lower()

user_continent = input(
    "Континент (Европа, Азия, Африка, Америка, Океания, любой): "
).strip().lower()

user_month = int(input("Месяц путешествия (1–12): "))

user_climate = input(
    "Климат (жаркий, тёплый, умеренный, прохладный): "
).strip().lower()

user_interests = input(
    "Интересы через запятую (море, история, природа, развлечения, гастрономия, спорт): "
)

user_interests = [
    interest.strip().lower()
    for interest in user_interests.split(",")
    if interest.strip()
]

user_interests = list(dict.fromkeys(user_interests))

user_rest = input(
    "Тип отдыха (спокойный, активный, смешанный): "
).strip().lower()

# Получаем рекомендации
recommendations = recommend_countries(
    user_budget,
    user_continent,
    user_month,
    user_climate,
    user_interests,
    user_rest
)

# Выводим результат
print("\nТОП-5 рекомендаций Travel Compass:\n")

for i, result in enumerate(recommendations, start=1):

    print(i, ".", result["country"], "—", result["score"], "баллов")

    if result["reasons"]:
        print("Причины:", ", ".join(result["reasons"]))
    else:
        print("Совпадений по оцениваемым критериям нет")

    print()


```
